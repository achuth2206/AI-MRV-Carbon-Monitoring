```python
import numpy as np
import tensorflow as tf
from utils import load_tiff_image

IMG_SIZE = 128
MODEL_PATH = "mangrove_classifier.h5"

# Load trained model
model = tf.keras.models.load_model(MODEL_PATH)

def predict_image(image_path):
    # Load and preprocess image
    img = load_tiff_image(image_path, (IMG_SIZE, IMG_SIZE))
    img = img / 255.0
    img = np.expand_dims(img, axis=0)  # Add batch dimension

    # Predict
    prediction = model.predict(img)[0][0]

    if prediction > 0.5:
        print(f"Prediction: Mangroves 🌿 ({prediction*100:.2f}% confidence)")
    else:
        print(f"Prediction: Non-Mangroves 🌍 ({(1-prediction)*100:.2f}% confidence)")

# 🔹 Change this path to your test image
test_image_path = "../dataset/NonMangroves/NonMangrove184.tif"

predict_image(test_image_path)
```
